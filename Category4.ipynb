```python
from collections import deque

def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)

            for neighbour in graph[node]:
                new_path = path + [neighbour]
                queue.append(new_path)

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

start = 'A'
goal = 'F'

path = bfs(graph, start, goal)

print("BFS Shortest Path:", path)
```

```python
def dfs(graph, node, goal, visited=None):
    if visited is None:
        visited = set()

    if node == goal:
        return [node]

    visited.add(node)

    for neighbour in graph[node]:
        if neighbour not in visited:
            path = dfs(graph, neighbour, goal, visited)

            if path:
                return [node] + path

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

start = 'A'
goal = 'F'

path = dfs(graph, start, goal)

print("DFS Path:", path)
```

```python
def depth_limited_search(graph, node, goal, limit, path=None):
    if path is None:
        path = [node]

    if node == goal:
        return path

    if limit == 0:
        return None

    for neighbour in graph[node]:
        if neighbour not in path:
            result = depth_limited_search(
                graph,
                neighbour,
                goal,
                limit - 1,
                path + [neighbour]
            )

            if result:
                return result

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': ['G'],
    'E': [],
    'F': [],
    'G': []
}

start = 'A'
goal = 'G'
limit = 3

path = depth_limited_search(graph, start, goal, limit)

if path:
    print("DLS Path:", path)
else:
    print("Goal not found within depth limit")
```

```python
from collections import deque

def bidirectional_search(graph, start, goal):

    if start == goal:
        return [start]

    front_queue = deque([[start]])
    back_queue = deque([[goal]])

    front_visited = {start: [start]}
    back_visited = {goal: [goal]}

    while front_queue and back_queue:

        # Search from start
        path = front_queue.popleft()
        node = path[-1]

        for neighbour in graph[node]:
            if neighbour not in front_visited:
                front_visited[neighbour] = path + [neighbour]
                front_queue.append(path + [neighbour])

                if neighbour in back_visited:
                    return (front_visited[neighbour] +
                            back_visited[neighbour][-2::-1])

        # Search from goal
        path = back_queue.popleft()
        node = path[-1]

        for neighbour in graph[node]:
            if neighbour not in back_visited:
                back_visited[neighbour] = path + [neighbour]
                back_queue.append(path + [neighbour])

                if neighbour in front_visited:
                    return (front_visited[neighbour] +
                            back_visited[neighbour][-2::-1])

    return None


graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'E'],
    'D': ['B', 'F'],
    'E': ['C', 'F'],
    'F': ['D', 'E']
}

start = 'A'
goal = 'F'

path = bidirectional_search(graph, start, goal)

print("Bidirectional Search Path:", path)
```
