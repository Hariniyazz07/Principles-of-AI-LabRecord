Greedy first search for 8-puzzel


```python
import heapq

def heuristic(state, goal):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count

def get_moves(state):
    moves = []
    pos = state.index(0)
    row = pos // 3
    col = pos % 3

    if row > 0:
        new = list(state)
        new[pos], new[pos-3] = new[pos-3], new[pos]
        moves.append(tuple(new))

    if row < 2:
        new = list(state)
        new[pos], new[pos+3] = new[pos+3], new[pos]
        moves.append(tuple(new))

    if col > 0:
        new = list(state)
        new[pos], new[pos-1] = new[pos-1], new[pos]
        moves.append(tuple(new))

    if col < 2:
        new = list(state)
        new[pos], new[pos+1] = new[pos+1], new[pos]
        moves.append(tuple(new))

    return moves

def greedy(start, goal):
    queue = []
    heapq.heappush(queue, (heuristic(start, goal), start))

    visited = set()
    parent = {start: None}

    while queue:
        h, current = heapq.heappop(queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]
            return path[::-1]

        for next_state in get_moves(current):
            if next_state not in visited:
                if next_state not in parent:
                    parent[next_state] = current
                heapq.heappush(queue, (heuristic(next_state, goal), next_state))

    return None

start = (1, 2, 3, 4, 0, 6, 7, 5, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

path = greedy(start, goal)

if path:
    for state in path:
        print(state[0:3])
        print(state[3:6])
        print(state[6:9])
        print()
else:
    print("No solution")
```

```python
import heapq

def heuristic(state, goal):
    distance = 0

    for i in range(9):
        if state[i] == 0:
            continue

        pos = goal.index(state[i])

        r1 = i // 3
        c1 = i % 3

        r2 = pos // 3
        c2 = pos % 3

        distance += abs(r1 - r2) + abs(c1 - c2)

    return distance

def get_moves(state):
    moves = []
    pos = state.index(0)

    row = pos // 3
    col = pos % 3

    if row > 0:
        new = list(state)
        new[pos], new[pos-3] = new[pos-3], new[pos]
        moves.append(tuple(new))

    if row < 2:
        new = list(state)
        new[pos], new[pos+3] = new[pos+3], new[pos]
        moves.append(tuple(new))

    if col > 0:
        new = list(state)
        new[pos], new[pos-1] = new[pos-1], new[pos]
        moves.append(tuple(new))

    if col < 2:
        new = list(state)
        new[pos], new[pos+1] = new[pos+1], new[pos]
        moves.append(tuple(new))

    return moves

def a_star(start, goal):
    queue = []
    heapq.heappush(queue, (heuristic(start, goal), 0, start))

    cost = {start: 0}
    parent = {start: None}

    while queue:
        f, g, current = heapq.heappop(queue)

        if current == goal:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        for next_state in get_moves(current):
            new_cost = g + 1

            if next_state not in cost or new_cost < cost[next_state]:
                cost[next_state] = new_cost
                parent[next_state] = current

                h = heuristic(next_state, goal)
                f = new_cost + h

                heapq.heappush(queue, (f, new_cost, next_state))

    return None

start = (1, 2, 3, 4, 0, 6, 7, 5, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

path = a_star(start, goal)

for state in path:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```

Memory Bound Heuristic Search - 8 puzzel


```python
def heuristic(state, goal):
    distance = 0

    for i in range(9):
        if state[i] == 0:
            continue

        pos = goal.index(state[i])

        r1 = i // 3
        c1 = i % 3

        r2 = pos // 3
        c2 = pos % 3

        distance += abs(r1 - r2) + abs(c1 - c2)

    return distance

def moves(state):
    result = []

    pos = state.index(0)
    row = pos // 3
    col = pos % 3

    if row > 0:
        new = list(state)
        new[pos], new[pos - 3] = new[pos - 3], new[pos]
        result.append(tuple(new))

    if row < 2:
        new = list(state)
        new[pos], new[pos + 3] = new[pos + 3], new[pos]
        result.append(tuple(new))

    if col > 0:
        new = list(state)
        new[pos], new[pos - 1] = new[pos - 1], new[pos]
        result.append(tuple(new))

    if col < 2:
        new = list(state)
        new[pos], new[pos + 1] = new[pos + 1], new[pos]
        result.append(tuple(new))

    return result

def search(state, goal, g, limit, path):
    f = g + heuristic(state, goal)

    if f > limit:
        return f

    if state == goal:
        return path

    minimum = 9999

    for next_state in moves(state):
        if next_state not in path:
            path.append(next_state)

            result = search(
                next_state,
                goal,
                g + 1,
                limit,
                path
            )

            if isinstance(result, list):
                return result

            if result < minimum:
                minimum = result

            path.pop()

    return minimum

def ida_star(start, goal):
    limit = heuristic(start, goal)
    path = [start]

    while True:
        result = search(
            start,
            goal,
            0,
            limit,
            path
        )

        if isinstance(result, list):
            return result

        limit = result

start = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

path = ida_star(start, goal)

for state in path:
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()
```

```python
import heapq

def valid(m, c):
    mr = 3 - m
    cr = 3 - c

    if m < 0 or c < 0 or m > 3 or c > 3:
        return False

    if m > 0 and m < c:
        return False

    if mr > 0 and mr < cr:
        return False

    return True

def moves(state):
    m, c, boat = state
    result = []

    possible = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

    for x, y in possible:
        if boat == 0:
            new = (m - x, c - y, 1)
        else:
            new = (m + x, c + y, 0)

        if valid(new[0], new[1]):
            result.append(new)

    return result

def heuristic(state):
    return state[0] + state[1]

def a_star(start, goal):
    queue = []
    heapq.heappush(queue, (heuristic(start), 0, start))

    cost = {start: 0}
    parent = {start: None}

    while queue:
        f, g, current = heapq.heappop(queue)

        if current == goal:
            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            return path[::-1]

        for next_state in moves(current):
            new_cost = g + 1

            if next_state not in cost or new_cost < cost[next_state]:
                cost[next_state] = new_cost
                parent[next_state] = current

                f = new_cost + heuristic(next_state)
                heapq.heappush(queue, (f, new_cost, next_state))

    return None

start = (3, 3, 0)
goal = (0, 0, 1)

path = a_star(start, goal)

for state in path:
    print(state)
```

AO* Search

```python
graph = {
    'A': [[('B', 1), ('C', 1)], [('D', 1)]],
    'B': [[('E', 1)]],
    'C': [[('F', 1)]],
    'D': [[('G', 1), ('H', 1)]],
    'E': [],
    'F': [],
    'G': [],
    'H': []
}

h = {
    'A': 4,
    'B': 2,
    'C': 2,
    'D': 3,
    'E': 0,
    'F': 0,
    'G': 0,
    'H': 0
}

solution = {}

def ao_star(node):
    if len(graph[node]) == 0:
        return h[node]

    min_cost = 9999
    best = None

    for option in graph[node]:
        cost = 0

        for child, value in option:
            cost += value + ao_star(child)

        if cost < min_cost:
            min_cost = cost
            best = option

    h[node] = min_cost
    solution[node] = best

    return min_cost

cost = ao_star('A')

print("Cost =", cost)

for node in solution:
    print(node, "->", solution[node])
```
