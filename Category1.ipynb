```python
class SimpleReflexAgent:
    def act(self, location, status):
        if status == "Dirty":
            return "Suck"
        elif location == "A":
            return "Move Right"
        elif location == "B":
            return "Move Left"
        else:
            return "No Action"
agent = SimpleReflexAgent()
test_cases = [
    ("A", "Dirty"),
    ("A", "Clean"),
    ("B", "Dirty"),
    ("B", "Clean")
]
for location, status in test_cases:
    action = agent.act(location, status)
    print(f"Location: {location}, Status: {status} --> Action: {action}")
```

```python
class ModelBasedAgent:
    def __init__(self):
        self.model = {
            "A": "Unknown",
            "B": "Unknown"
        }
    def act(self, location, status):
        self.model[location] = status
        if status == "Dirty":
            return "Suck"
        if self.model["A"] == "Clean" and self.model["B"] == "Clean":
            return "Stop"
        if location == "A":
            return "Move Right"
        if location == "B":
            return "Move Left"
agent = ModelBasedAgent()
percepts = [
    ("A", "Dirty"),
    ("A", "Clean"),
    ("B", "Dirty"),
    ("B", "Clean")
]
for location, status in percepts:
    action = agent.act(location, status)
    print(f"Location: {location}, Status: {status}")
    print(f"Action: {action}")
    print(f"Model: {agent.model}")
    print("-" * 30)
```

```python
class GoalBasedAgent:
    def __init__(self):
        self.goal = "All rooms clean"
    def act(self, rooms):
        if all(status == "Clean" for status in rooms.values()):
            return "Goal Achieved! Stop."
        for room, status in rooms.items():
            if status == "Dirty":
                rooms[room] = "Clean"
                return f"Cleaning Room {room}"
rooms = {
    "A": "Dirty",
    "B": "Dirty"
}
agent = GoalBasedAgent()
while True:
    action = agent.act(rooms)
    print("Action:", action)
    print("Rooms:", rooms)
    if action == "Goal Achieved! Stop.":
        break
```

```python
class UtilityBasedAgent:
    def __init__(self):
        self.utilities = {
            "Clean": 10,
            "Move": 5,
            "Stop": 0
        }
    def act(self, location, status):
        if status == "Dirty":
            action = "Clean"
        else:
            if location == "A":
                action = "Move to B"
            elif location == "B":
                action = "Move to A"
            else:
                action = "Stop"

        return action
agent = UtilityBasedAgent()
test_cases = [
    ("A", "Dirty"),
    ("A", "Clean"),
    ("B", "Dirty"),
    ("B", "Clean")
]
for location, status in test_cases:
    action = agent.act(location, status)
    print(f"Location: {location}, Status: {status} --> Action: {action}")
```
