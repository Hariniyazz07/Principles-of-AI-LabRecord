Min-Max Search algorithm for chess game


```python
def minimax(board, depth, maximizing):
    if depth == 0:
        return evaluate(board)

    if maximizing:
        best = -9999

        for move in get_moves(board, 'W'):
            new_board = make_move(board, move)
            value = minimax(new_board, depth - 1, False)
            best = max(best, value)

        return best

    else:
        best = 9999

        for move in get_moves(board, 'B'):
            new_board = make_move(board, move)
            value = minimax(new_board, depth - 1, True)
            best = min(best, value)

        return best


def evaluate(board):
    values = {
        'P': 1,
        'N': 3,
        'B': 3,
        'R': 5,
        'Q': 9,
        'K': 100
    }

    score = 0

    for row in board:
        for piece in row:
            if piece.isupper():
                score += values.get(piece.upper(), 0)
            elif piece.islower():
                score -= values.get(piece.upper(), 0)

    return score


def get_moves(board, player):
    moves = []

    for i in range(8):
        for j in range(8):
            if player == 'W' and board[i][j].isupper():
                if i > 0:
                    moves.append((i, j, i - 1, j))

            elif player == 'B' and board[i][j].islower():
                if i < 7:
                    moves.append((i, j, i + 1, j))

    return moves


def make_move(board, move):
    new_board = [row[:] for row in board]

    x1, y1, x2, y2 = move
    new_board[x2][y2] = new_board[x1][y1]
    new_board[x1][y1] = '.'

    return new_board


board = [
    ['r', 'n', 'b', 'q', 'k', 'b', 'n', 'r'],
    ['p', 'p', 'p', 'p', 'p', 'p', 'p', 'p'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['P', 'P', 'P', 'P', 'P', 'P', 'P', 'P'],
    ['R', 'N', 'B', 'Q', 'K', 'B', 'N', 'R']
]

print("Min-Max value:", minimax(board, 2, True))
```

Alpha-Beta Pruning Algorithm for Chess Game



```python
def alpha_beta(board, depth, alpha, beta, maximizing):
    if depth == 0:
        return evaluate(board)

    if maximizing:
        best = -9999

        for move in get_moves(board, 'W'):
            new_board = make_move(board, move)
            value = alpha_beta(new_board, depth - 1, alpha, beta, False)

            best = max(best, value)
            alpha = max(alpha, best)

            if beta <= alpha:
                break

        return best

    else:
        best = 9999

        for move in get_moves(board, 'B'):
            new_board = make_move(board, move)
            value = alpha_beta(new_board, depth - 1, alpha, beta, True)

            best = min(best, value)
            beta = min(beta, best)

            if beta <= alpha:
                break

        return best


def evaluate(board):
    values = {
        'P': 1,
        'N': 3,
        'B': 3,
        'R': 5,
        'Q': 9,
        'K': 100
    }

    score = 0

    for row in board:
        for piece in row:
            if piece.isupper():
                score += values.get(piece.upper(), 0)
            elif piece.islower():
                score -= values.get(piece.upper(), 0)

    return score


def get_moves(board, player):
    moves = []

    for i in range(8):
        for j in range(8):
            if player == 'W' and board[i][j].isupper():
                if i > 0:
                    moves.append((i, j, i - 1, j))

            elif player == 'B' and board[i][j].islower():
                if i < 7:
                    moves.append((i, j, i + 1, j))

    return moves


def make_move(board, move):
    new_board = [row[:] for row in board]

    x1, y1, x2, y2 = move
    new_board[x2][y2] = new_board[x1][y1]
    new_board[x1][y1] = '.'

    return new_board


board = [
    ['r', 'n', 'b', 'q', 'k', 'b', 'n', 'r'],
    ['p', 'p', 'p', 'p', 'p', 'p', 'p', 'p'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['.', '.', '.', '.', '.', '.', '.', '.'],
    ['P', 'P', 'P', 'P', 'P', 'P', 'P', 'P'],
    ['R', 'N', 'B', 'Q', 'K', 'B', 'N', 'R']
]

result = alpha_beta(board, 2, -9999, 9999, True)

print("Alpha-Beta value:", result)
```

Min-Max Search Algorithm for Tic-Tac-Toe

```python
def print_board(board):
    print()
    for row in board:
        print(" | ".join(row))
        print("-" * 9)


def check_winner(board):
    for i in range(3):
        if board[i][0] == board[i][1] == board[i][2] != ' ':
            return board[i][0]

        if board[0][i] == board[1][i] == board[2][i] != ' ':
            return board[0][i]

    if board[0][0] == board[1][1] == board[2][2] != ' ':
        return board[0][0]

    if board[0][2] == board[1][1] == board[2][0] != ' ':
        return board[0][2]

    return None


def is_full(board):
    for row in board:
        if ' ' in row:
            return False
    return True


def minimax(board, maximizing):
    winner = check_winner(board)

    if winner == 'O':
        return 1

    if winner == 'X':
        return -1

    if is_full(board):
        return 0

    if maximizing:
        best = -999

        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    board[i][j] = 'O'
                    value = minimax(board, False)
                    board[i][j] = ' '
                    best = max(best, value)

        return best

    else:
        best = 999

        for i in range(3):
            for j in range(3):
                if board[i][j] == ' ':
                    board[i][j] = 'X'
                    value = minimax(board, True)
                    board[i][j] = ' '
                    best = min(best, value)

        return best


def best_move(board):
    best_value = -999
    move = None

    for i in range(3):
        for j in range(3):
            if board[i][j] == ' ':
                board[i][j] = 'O'
                value = minimax(board, False)
                board[i][j] = ' '

                if value > best_value:
                    best_value = value
                    move = (i, j)

    return move


board = [
    [' ', ' ', ' '],
    [' ', ' ', ' '],
    [' ', ' ', ' ']
]

while True:
    print_board(board)

    row = int(input("Enter row (1-3): ")) - 1
    col = int(input("Enter column (1-3): ")) - 1

    if board[row][col] != ' ':
        print("Invalid move")
        continue

    board[row][col] = 'X'

    if check_winner(board) == 'X':
        print_board(board)
        print("You win!")
        break

    if is_full(board):
        print_board(board)
        print("Draw!")
        break

    move = best_move(board)
    board[move[0]][move[1]] = 'O'

    print("Computer played:", move[0] + 1, move[1] + 1)

    if check_winner(board) == 'O':
        print_board(board)
        print("Computer wins!")
        break

    if is_full(board):
        print_board(board)
        print("Draw!")
        break
```
