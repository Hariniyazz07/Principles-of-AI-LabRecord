```python
# Fully Observable
rooms = {
    "A": "Dirty",
    "B": "Clean"
}
for i in rooms:
    if rooms[i] == "Dirty":
        print(i, "is Dirty")
        print("Cleaning", i)
    else:
        print(i, "is already Clean")
print("\nPartially Observable")
# Only Room A can be seen
room = {
    "A": "Dirty"
}
for i in room:
    if room[i] == "Dirty":
        print("Cleaning", i)
```

```python
print("Known Environment")
rooms = ["Dirty", "Clean"]
for i in rooms:
    print(i)
print("\nUnknown Environment")
rooms = []
if len(rooms) == 0:
    print("No information about environment")
```

```python
import random
print("Static")
state = "Dirty"
if state == "Dirty":
    print("Cleaning Room")
print("\nDynamic")
state = random.choice(["Dirty", "Clean"])
print("Current State :", state)
if state == "Dirty":
    print("Cleaning Room")
else:
    print("Nothing to do")
```

```python
print("Discrete")
room = "Dirty"
if room == "Dirty":
    print("Clean the room")
print("\nContinuous")
temp = float(input("Enter Temperature: "))
if temp > 30:
    print("Switch ON Fan")
else:
    print("Switch OFF Fan")
```

```python
import random
print("Deterministic")
room = "Dirty"
if room == "Dirty":
    room = "Clean"
print(room)
print("\nStochastic")
room = "Dirty"
chance = random.randint(1, 2)
if chance == 1:
    room = "Clean"
    print("Cleaning Successful")
else:
    print("Cleaning Failed")
print(room)
```

```python
print("Episodic")
rooms = ["Dirty", "Clean", "Dirty"]
for i in rooms:
    if i == "Dirty":
        print("Clean")
    else:
        print("Skip")
print("\nSequential")
battery = 3
while battery > 0:
    print("Cleaning Room")
    battery = battery - 1
print("Battery Empty")
```
