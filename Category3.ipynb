```python
graph = {
    0: [1, 2],
    1: [3, 4],
    2: [5, 6],
    3: [7],
    4: [8],
    5: [9],
    6: [10],
    7: [],
    8: [],
    9: [],
    10: []
}
value = {
    0: 10,
    1: 20,
    2: 15,
    3: 25,
    4: 30,
    5: 18,
    6: 22,
    7: 28,
    8: 30,
    9: 35,
    10: 100
}
def local_beam_search(start, k):
    states = [start]
    while True:
        successors = []
        for state in states:
            successors += graph[state]
        if not successors:
            break
        successors.sort(key=lambda x: value[x], reverse=True)
        new_states = successors[:k]
        if max(value[x] for x in new_states) <= max(value[x] for x in states):
            break
        states = new_states
    best = max(states, key=lambda x: value[x])
    return best, value[best]
start = 0
for k in [1, 2, 3]:
    state, score = local_beam_search(start, k)
    print("Beam Width:", k)
    print("Best State:", state)
    print("Value:", score)
    print()
global_state = max(value, key=value.get)
print("Global Optimum:", global_state)
print("Global Value:", value[global_state])
```

```python
def conflicts(a):
    c = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if a[i] == a[j] or abs(a[i] - a[j]) == abs(i - j):
                c += 1
    return c
def hill_climbing():
    board = [0, 4, 7, 5, 2, 6, 1, 3]
    while True:
        current = conflicts(board)
        if current == 0:
            return board
        best = board
        best_score = current
        for i in range(8):
            for j in range(8):
                new = board[:]
                new[i] = j
                score = conflicts(new)
                if score < best_score:
                    best = new
                    best_score = score
        if best_score == current:
            return board
        board = best
solution = hill_climbing()
print("Solution:", solution)
for i in range(8):
    for j in range(8):
        if solution[i] == j:
            print("Q", end=" ")
        else:
            print(".", end=" ")
    print()
```

```python
import random
import math

def find_value(x):
    return -(x - 5) ** 2 + 25

x = random.randint(0, 10)
temp = 100

while temp > 0.1:
    new_x = x + random.choice([-1, 1])

    if new_x >= 0 and new_x <= 10:
        old_value = find_value(x)
        new_value = find_value(new_x)

        if new_value > old_value:
            x = new_x
        else:
            p = math.exp((new_value - old_value) / temp)

            if random.random() < p:
                x = new_x

    temp = temp * 0.9

print("Best solution:", x)
print("Best value:", find_value(x))
```
