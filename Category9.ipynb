Medical Diagnosis using Bayesian Reasoning

```python
disease = float(input("Enter probability of disease (0-1): "))
sensitivity = float(input("Enter test sensitivity (0-1): "))
false_positive = float(input("Enter false positive rate (0-1): "))
positive = input("Is the test positive? (yes/no): ").lower()
if positive == "yes":
    numerator = sensitivity * disease
    denominator = numerator + false_positive * (1 - disease)
else:
    numerator = (1 - sensitivity) * disease
    denominator = numerator + (1 - false_positive) * (1 - disease)
if denominator != 0:
    result = numerator / denominator
    print("Probability of disease:", round(result * 100, 2), "%")
else:
    print("Probability cannot be calculated")
```

Electric and Autonomous Vehicle using Bayesian Reasoning

```python
prior = 0.7
camera_likelihood = 0.9
other_likelihood = 0.2
numerator = prior * camera_likelihood
denominator = numerator + (1 - prior) * other_likelihood
probability = numerator / denominator
print("Probability of pedestrian:",
      round(probability * 100, 2), "%")
if probability > 0.5:
    print("Decision: Apply emergency braking")
else:
    print("Decision: Continue monitoring")
```

Smart Manufacturing using Bayesian Reasoning

```python
prior = 0.1
temp_likelihood = 0.8
vibration_likelihood = 0.9
normal_temp = 0.2
normal_vibration = 0.1
numerator = prior * temp_likelihood * vibration_likelihood
denominator = numerator + (
    (1 - prior) * normal_temp * normal_vibration
)
probability = numerator / denominator
print("Machine Failure Probability:",
      round(probability * 100, 2), "%")
if probability > 0.5:
    print("Alert: Machine maintenance required")
else:
    print("Machine condition: Normal")
```

Medical Diagnosis using Fuzzy Logic

```python
temperature = float(input("Enter body temperature: "))
if temperature < 37:
    risk = "LOW"
elif temperature < 39:
    risk = "MEDIUM"
else:
    risk = "HIGH"
print("Patient Temperature:", temperature)
print("Medical Risk Level:", risk)
```

Electric and Autonomous Vehicle using Fuzzy Logic

```python
distance = float(input("Enter obstacle distance in meters: "))
if distance <= 10:
    braking = "HIGH"
elif distance <= 30:
    braking = "MEDIUM"
else:
    braking = "LOW"
print("Obstacle Distance:", distance, "meters")
print("Braking Level:", braking)
```

Smart Manufacturing using Fuzzy Logic

```python
temperature = float(input("Enter machine temperature: "))
if temperature < 40:
    cooling = "LOW"
elif temperature < 70:
    cooling = "MEDIUM"
else:
    cooling = "HIGH"
print("Machine Temperature:", temperature)
print("Cooling Level:", cooling)
if cooling == "HIGH":
    print("Activate maximum cooling")
elif cooling == "MEDIUM":
    print("Activate moderate cooling")
else:
    print("Normal cooling")
```
